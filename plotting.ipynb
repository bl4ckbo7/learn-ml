```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df = pd.read_csv('titanic.csv')
plt.scatter(df['Age'], df['Fare']) #plt.scatter(x,y)

plt.xlabel('Age') # add x-axis label
plt.ylabel('Fare') # add y-axis label

plt.scatter(df['Age'], df['Fare'], c=df['Pclass']) #Color coding it according to Passenger class. We add the c parameter and give it a Pandas series. In this case, our Pandas series has 3 possible values (1st, 2nd, and 3rd class), so we'll see our datapoints each get one of three colors.
```

```python
age_arr = df['Age'].values #put the values in an array
fare_arr = df['Fare'].values
print(np.count_nonzero(age_arr == 80)) #count how many passengers are 80 years old
```

```python
# How many passengers who are aged 80 years old had paid fare of less than 100?
v = df[['Age','Fare']].values
age80 = v[:,0]==80
fare100 = v[:,1] < 100
combined = age80 & fare100 # Bitwise AND (&)
print(v[combined]) #array of result
print(np.count_nonzero(v[combined][:,0])) #count
```

```python
plt.scatter(df['Age'], df['Fare'], c=df['Pclass']) #The following draws a line to approximately separate the 1st class from the 2nd and 3rd class. From eyeballing, we’ll put the line from (0, 85) to (80, 5).
plt.plot([0, 80],[85, 5])
plt.show()
```
