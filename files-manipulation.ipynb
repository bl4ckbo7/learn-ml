```python
arr = df[['Pclass','Fare','Age']]
print(arr.shape)   #'shape' attribute returns size or the total number of rows and columns (n_rows,n_columns). Also,
print(df.shape) #return size of the full dataFrame

arr_vals = arr.values

#select a single row from the numpy array
print(arr_vals[0])

#select a single cell from the numpy array
print(arr_vals[0,1])

#select single column only
print(arr_vals[:,1])

```

```python
# Masking

#select all the rows of children (passengers under the age of 18
mask = arr_vals[:,2] < 18
print(mask)
# print(arr_vals[mask])
#now just return the numerical values instead of boolean
print(arr_vals[mask,1])

#how many of them are under 18? Count the booleans TRUE=1, FALSE=0.
print(mask.sum()) #130
```
